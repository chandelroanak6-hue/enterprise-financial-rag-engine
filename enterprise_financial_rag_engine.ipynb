```python
import os
import pandas as pd
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

# 1. Simulate an Enterprise Financial Disclosure (10-K / Earnings Excerpt)
financial_filing_corpus = """
Alphabet Inc. announced its financial results for the quarter ended March 31.
Consolidated revenues were $80.5 billion, reflecting an increase of 15% year-over-year.
Operating income reached $25.5 billion, representing an operating margin of 32%, up from 25% in the prior year period.
Google Cloud delivered segment revenue of $9.57 billion, growing 28% year-over-year, driven by enterprise adoption of Google Cloud Platform (GCP) and infrastructure services.
Operating income for Google Cloud surged to $900 million, compared to an operating loss of $186 million in the same quarter last year, reflecting strong unit economics and cost optimization.
Capital expenditures (CapEx) for the quarter totaled $12.0 billion, primarily allocated toward technical infrastructure, including servers, datacenters, and custom TPU processors to support enterprise AI workloads.
Free Cash Flow for the quarter stood at $16.8 billion, maintaining robust enterprise liquidity.
"""

# 2. Advanced Text Chunking with Overlap (Preserves Financial Context)
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=200, chunk_overlap=40, separators=['\n', '. ', ' ']
)
chunks = text_splitter.split_text(financial_filing_corpus)

# 3. Generate Semantic Embeddings using a Hugging Face Transformer
embedding_engine = HuggingFaceEmbeddings(
    model_name='sentence-transformers/all-MiniLM-L6-v2'
)

# 4. Ingest Chunks into ChromaDB Vector Store
vector_db = Chroma.from_texts(
    texts=chunks, embedding=embedding_engine, collection_name='financial_10k'
)

print(
    f'Successfully vectorized and indexed {len(chunks)} financial chunks into ChromaDB.'
)
```

```python
def query_financial_agent(query: str, top_k: int = 2):
  # Perform cosine similarity search over embedded financial chunks
  results = vector_db.similarity_search_with_score(query, k=top_k)

  print(f"=== Analytical Query: '{query}' ===")
  for rank, (doc, score) in enumerate(results, 1):
    print(
        f'\n[Rank {rank} Source Evidence | Relevance Distance:'
        f' {round(score, 4)}]'
    )
    print(f'"{doc.page_content}"')


# Test with typical enterprise analyst questions:
query_financial_agent('What were Google Cloud revenues and profit figures?')
print('\n' + '=' * 50 + '\n')
query_financial_agent('What were the capital expenditure allocations and CapEx?')
```
